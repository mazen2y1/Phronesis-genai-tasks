SETUP


```python
from google import genai
from google.genai import types

API_key = "PUT YOUR API KEY HERE"

Client = genai.Client(api_key=API_key)
```

STEP 1

```python
content = input("ASK :")

response = Client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = content,
    config = types.GenerateContentConfig(
        system_instruction = "You are a instructor"
    )
)
print(response.text)
```

```python
print("RAW RESPONSE : \n")
print(response.text)
print("\n###########################################\n")
print("TOKEN USAGE :\n")
print(response.usage_metadata)
```

STEP 2


```python
content2=input("ASK : ")

response = Client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = content2,
    config=types.GenerateContentConfig(
        system_instruction=(
            "You are a strict technical writer. "
            "Always answer in exactly 3 bullet points, no more, no less. "
            "Each bullet must be a single short sentence."
        )
    )
)
print(response.text)
```

STEP 3

```python
few_shot_prompt = '''Convert the following product names into short marketing taglines.

Product: Wireless Earbuds
Tagline: Freedom for your ears.

Product: Standing Desk
Tagline: Work up, not out.

Product: Smart Water Bottle
Tagline:'''

response = Client.models.generate_content(
    model="gemini-2.5-flash",
    contents=few_shot_prompt
)
print(response.text)
```

STEP 4

```python
response = Client.models.generate_content(
    model="gemini-3.1-flash",
    contents="Give me one creative name for a coffee shop.",
    config=types.GenerateContentConfig(
        temperature=1.2,
        max_output_tokens=30, 
    )
)
print(response.text)
```

CHAT WITH BOT


```python
chat = Client.chats.create(
    model="gemini-2.5-flash",
    config=types.GenerateContentConfig(
        system_instruction="You are a friendly, concise AI assistant helping a student in this lab."
    )
)

print("Chatbot ready! Type 'exit' or 'quit' to stop.\n")

while True:
    user_input = input("You : ")
    if user_input.strip().lower() in ("exit", "quit"):
        print("Bot: Goodbye! ")
        break

    response = chat.send_message(user_input)
    print(f"Bot: {response.text}\n")
```
